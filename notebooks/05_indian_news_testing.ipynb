```python
from app.predictor import predict_news
import joblib

model = joblib.load("../models/linear_svm_model.pkl")
tfidf = joblib.load("../models/tfidf_vectorizer.pkl")
```

```python
article_test = article

prediction = predict_news(article_test)

text = predict_news.__globals__["clean_text"](article_test)
text = predict_news.__globals__["remove_stopwords"](text)
text = predict_news.__globals__["lemmatize_text"](text)

article_tfidf = tfidf.transform([text])

score = model.decision_function(article_tfidf)[0]

print("Decision score:", score)
print("Prediction:", prediction)
```

```python
for i, test_article in enumerate(
    [article1, article2, article3, article4], start=2
):
    text = predict_news.__globals__["clean_text"](test_article)
    text = predict_news.__globals__["remove_stopwords"](text)
    text = predict_news.__globals__["lemmatize_text"](text)

    article_tfidf = tfidf.transform([text])
    score = model.decision_function(article_tfidf)[0]
    prediction = model.predict(article_tfidf)[0]

    print(f"Article {i}:")
    print(f"  Decision score: {score:.4f}")
    print(f"  Prediction: {'REAL' if prediction == 1 else 'FAKE'}")
    print()
```
