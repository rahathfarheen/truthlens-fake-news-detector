```python
df = pd.read_csv("../data/processed_news.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
print(df["content"].iloc[0])
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    text = re.sub(r"<.*?>", "", text)
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
sample_text = df["content"].iloc[0]

print("BEFORE:")
print(sample_text)

print("\nAFTER:")
print(clean_text(sample_text))
```

```python
df["clean_text"] = df["content"].apply(clean_text)
```

```python
df[["content", "clean_text"]].head()
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

print("Number of stopwords:", len(stop_words))
print(list(stop_words)[:20])
```

```python
def remove_stopwords(text):
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return " ".join(words)
```

```python
sample = df["clean_text"].iloc[0]

print("BEFORE:")
print(sample)

print("\nAFTER:")
print(remove_stopwords(sample))
```

```python
df["no_stopwords"] = df["clean_text"].apply(remove_stopwords)
```

```python
df[["clean_text", "no_stopwords"]].head()
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
def lemmatize_text(text):
    words = text.split()
    words = [lemmatizer.lemmatize(word) for word in words]
    return " ".join(words)
```

```python
sample = df["no_stopwords"].iloc[0]

print("BEFORE:")
print(sample)

print("\nAFTER:")
print(lemmatize_text(sample))
```

```python
df["processed_text"] = df["no_stopwords"].apply(lemmatize_text)

print("Lemmatization completed!")
```

```python
display(df[["no_stopwords", "processed_text"]].head())
```

```python
sample = df["content"].iloc[0]

print("ORIGINAL:")
print(sample)

print("\n" + "="*80 + "\n")

print("PROCESSED:")
print(df["processed_text"].iloc[0])
```
